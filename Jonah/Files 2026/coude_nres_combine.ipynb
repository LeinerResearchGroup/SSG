```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from astropy.time import Time

base_dir = r"starstxt4"

rows_K = []   # semi-amplitude table if you want it later

K1_val, K2_val = [], []
K1_upper, K2_upper, K1_lower, K2_lower = [], [], [], []
star_names = []

outdir = "PNGs"
os.makedirs(outdir, exist_ok=True)

for star in sorted(os.listdir(base_dir)):
    star_dir = os.path.join(base_dir, star)
    if not os.path.isdir(star_dir):
        continue

    rv_path = os.path.join(star_dir, f"{star}_RVs.txt")
    if not os.path.isfile(rv_path):
        continue

    time_list   = []
    rv1_list    = []
    rv1_up_list = []
    rv1_low_list= []
    rv2_list    = []
    rv2_up_list = []
    rv2_low_list= []

    # Read file: for each line, always append a value for RV2 (NaN if none)
    with open(rv_path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            line = line.strip("[],")
            parts = [p.strip() for p in line.split(",")]

            # primary
            time     = float(parts[0])
            rv1      = float(parts[1])
            rv1_up   = float(parts[2])
            rv1_low  = float(parts[3])

            time_list.append(time)
            rv1_list.append(rv1)
            rv1_up_list.append(rv1_up)
            rv1_low_list.append(rv1_low)

            # default secondary as NaN
            rv2      = np.nan
            rv2_up   = np.nan
            rv2_low  = np.nan

            # secondary (optional)
            if len(parts) >= 7:
                rv2_s, rv2_up_s, rv2_low_s = parts[4], parts[5], parts[6]
                if rv2_s.lower() != "nan":
                    rv2     = float(rv2_s)
                    rv2_up  = float(rv2_up_s)
                    rv2_low = float(rv2_low_s)

            rv2_list.append(rv2)
            rv2_up_list.append(rv2_up)
            rv2_low_list.append(rv2_low)

    if len(time_list) == 0:
        continue


    # Convert to numpy arrays and to datetimes
    bjd      = np.array(time_list)
    rv1      = np.array(rv1_list)
    rv1_hi   = np.array(rv1_up_list)   # upward error
    rv1_lo   = np.array(rv1_low_list)  # downward error
    rv2      = np.array(rv2_list)
    rv2_hi   = np.array(rv2_up_list)
    rv2_lo   = np.array(rv2_low_list)

    t     = Time(bjd, format='jd', scale='tdb')
    dates = t.to_datetime()


    # PRIMARY RV stats
    primary_amp    = (np.nanmax(rv1) - np.nanmin(rv1)) / 2.0
    primary_mean   = np.nanmean(rv1)
    primary_std    = np.nanstd(rv1)
    rv1_hi_max     = np.nanmax(rv1_hi)
    rv1_hi_mean    = np.nanmean(rv1_hi)
    rv1_lo_max     = np.nanmax(rv1_lo)
    rv1_lo_mean    = np.nanmean(rv1_lo)

    print(f"""{star} — Primary
  Semi-Amplitude  : {primary_amp:.3f} km/s
  Mean            : {primary_mean:.3f} km/s
  STD             : {primary_std:.3f} km/s
Primary Error Bars
  RV1_hi max      : {rv1_hi_max:.3f} km/s
  RV1_hi mean     : {rv1_hi_mean:.3f} km/s
  RV1_lo max      : {rv1_lo_max:.3f} km/s
  RV1_lo mean     : {rv1_lo_mean:.3f} km/s
""")

    # SECONDARY stats (if any non-NaN RV2)
    mask2 = ~np.isnan(rv2)

    if mask2.any():
        rv2_v      = rv2[mask2]
        rv2_lo_v   = rv2_lo[mask2]
        rv2_hi_v   = rv2_hi[mask2]
        dates_v    = dates[mask2]

        secondary_amp   = (np.nanmax(rv2_v) - np.nanmin(rv2_v)) / 2.0
        secondary_mean  = np.nanmean(rv2_v)
        secondary_std   = np.nanstd(rv2_v)
        rv2_hi_max      = np.nanmax(rv2_hi_v)
        rv2_hi_mean     = np.nanmean(rv2_hi_v)
        rv2_lo_max      = np.nanmax(rv2_lo_v)
        rv2_lo_mean     = np.nanmean(rv2_lo_v)

        print(f"""{star} — Secondary
  Semi-Amplitude  : {secondary_amp:.3f} km/s
  Mean            : {secondary_mean:.3f} km/s
  STD             : {secondary_std:.3f} km/s
Secondary Error Bars
  RV2_hi max      : {rv2_hi_max:.3f} km/s
  RV2_hi mean     : {rv2_hi_mean:.3f} km/s
  RV2_lo max      : {rv2_lo_max:.3f} km/s
  RV2_lo mean     : {rv2_lo_mean:.3f} km/s
""")
    else:
        secondary_amp   = np.nan
        secondary_mean  = np.nan
        secondary_std   = np.nan
        print(f"{star} — Secondary RV2 is all NaN; skipping its stats and plot.\n")

    # Save K table info (optional)
    star_names.append(star)
    K1_val.append(primary_amp)
    K1_upper.append(rv1_hi_mean)
    K1_lower.append(rv1_lo_mean)

    K2_val.append(secondary_amp)
    if mask2.any():
        K2_upper.append(rv2_hi_mean)
        K2_lower.append(rv2_lo_mean)
    else:
        K2_upper.append(np.nan)
        K2_lower.append(np.nan)

    rows_K.append({
        "star": star,
        "K1": primary_amp,
        "K1_err_hi_mean": rv1_hi_mean,
        "K1_err_lo_mean": rv1_lo_mean,
        "K2": secondary_amp,
        "K2_err_hi_mean": rv2_hi_mean if mask2.any() else np.nan,
        "K2_err_lo_mean": rv2_lo_mean if mask2.any() else np.nan
    })

    # Plot
    fig, ax = plt.subplots(figsize=(7, 4))

    # primary
    ax.errorbar(dates, rv1,
                yerr=[rv1_lo, rv1_hi],
                fmt='o',
                label=f"Primary\n(K={primary_amp:.2f} km/s)",
                ecolor='grey', capsize=1, elinewidth=0.5,
                ms=5, alpha=0.9)

    # secondary (only if we have valid points)
    if mask2.any():
        ax.errorbar(dates_v, rv2_v,
                    yerr=[rv2_lo_v, rv2_hi_v],
                    fmt='x',
                    label=f"Secondary\n(K={secondary_amp:.2f} km/s)",
                    ecolor='grey', capsize=1, elinewidth=0.5,
                    ms=5, alpha=0.9)
        ax.axhline(secondary_mean, alpha=0.25)

    # primary mean line
    ax.axhline(primary_mean, alpha=0.25)

    # y-axis padding
    if mask2.any():
        y_min = min(np.nanmin(rv1), np.nanmin(rv2_v)) - 5
        y_max = max(np.nanmax(rv1), np.nanmax(rv2_v)) + 5
    else:
        y_min = np.nanmin(rv1) - 5
        y_max = np.nanmax(rv1) + 5
    ax.set_ylim(y_min, y_max)

    # date formatting
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    fig.autofmt_xdate()

    ax.set_xlabel('Date (UTC)')
    ax.set_ylabel('Radial Velocity (km/s)')
    ax.set_title('RV vs Time for ' + star)

    # legend outside
    box = ax.get_position()
    ax.set_position([box.x0, box.y0, box.width * 0.75, box.height])
    ax.legend(loc='center left', bbox_to_anchor=(1, 0.5))

    plt.tight_layout()

    # save PNG per star
    outpath = os.path.join(outdir, f"{star}.png")
    #plt.savefig(outpath, dpi=135, bbox_inches='tight')
    plt.show()

    print(f"Number of points used for {star}: {len(bjd)}\n")

# Optional: assemble K semi-amplitude table as a DataFrame
df_K = pd.DataFrame({
    "Star": star_names,
    "K1": K1_val,
    "K1_err_hi_mean": K1_upper,
    "K1_err_lo_mean": K1_lower,
    "K2": K2_val,
    "K2_err_hi_mean": K2_upper,
    "K2_err_lo_mean": K2_lower
})

```

```python

# Combined plots for the 4 NRES stars (old starstxt4 + starstxt4_nres)


nres_base_dir = r"starstxt4_nres"
special_stars = ["29Dra", "AZPsc", "BD+201895", "DIPsc", "EZPeg", "V834Her"]

def load_rv_file(folder, star):
    """
    Load one star's RV file from a given folder.
    Returns dict with numpy arrays or None if file missing.
    """
    star_dir = os.path.join(folder, star)
    rv_path  = os.path.join(star_dir, f"{star}_RVs.txt")
    if not os.path.isfile(rv_path):
        return None

    time_list   = []
    rv1_list    = []
    rv1_up_list = []
    rv1_low_list= []
    rv2_list    = []
    rv2_up_list = []
    rv2_low_list= []

    with open(rv_path, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            line  = line.strip("[],")
            parts = [p.strip() for p in line.split(",")]

            # primary
            time    = float(parts[0])
            rv1     = float(parts[1])
            rv1_up  = float(parts[2])
            rv1_low = float(parts[3])

            time_list.append(time)
            rv1_list.append(rv1)
            rv1_up_list.append(rv1_up)
            rv1_low_list.append(rv1_low)

            # default secondary = NaN
            rv2     = np.nan
            rv2_up  = np.nan
            rv2_low = np.nan

            # secondary (optional)
            if len(parts) >= 7:
                rv2_s, rv2_up_s, rv2_low_s = parts[4], parts[5], parts[6]
                if rv2_s.lower() != "nan":
                    rv2     = float(rv2_s)
                    rv2_up  = float(rv2_up_s)
                    rv2_low = float(rv2_low_s)

            rv2_list.append(rv2)
            rv2_up_list.append(rv2_up)
            rv2_low_list.append(rv2_low)

    return {
        "bjd":   np.array(time_list),
        "rv1":   np.array(rv1_list),
        "rv1_hi":np.array(rv1_up_list),
        "rv1_lo":np.array(rv1_low_list),
        "rv2":   np.array(rv2_list),
        "rv2_hi":np.array(rv2_up_list),
        "rv2_lo":np.array(rv2_low_list),
    }

outdir_comb = "PNGs_combined"
os.makedirs(outdir_comb, exist_ok=True)

for star in special_stars:
    # load from both old and NRES folders (skip missing)
    datasets = []
    d_old  = load_rv_file(base_dir,      star)
    d_nres = load_rv_file(nres_base_dir, star)
    if d_old  is not None: datasets.append(d_old)
    if d_nres is not None: datasets.append(d_nres)

    if not datasets:
        print(f"{star}: no data found in either folder, skipping.")
        continue

    # concatenate/combine all data sets
    bjd    = np.concatenate([d["bjd"]    for d in datasets])
    rv1    = np.concatenate([d["rv1"]    for d in datasets])
    rv1_hi = np.concatenate([d["rv1_hi"] for d in datasets])
    rv1_lo = np.concatenate([d["rv1_lo"] for d in datasets])
    rv2    = np.concatenate([d["rv2"]    for d in datasets])
    rv2_hi = np.concatenate([d["rv2_hi"] for d in datasets])
    rv2_lo = np.concatenate([d["rv2_lo"] for d in datasets])

    # sort by time
    order  = np.argsort(bjd)
    bjd    = bjd[order]
    rv1    = rv1[order]
    rv1_hi = rv1_hi[order]
    rv1_lo = rv1_lo[order]
    rv2    = rv2[order]
    rv2_hi = rv2_hi[order]
    rv2_lo = rv2_lo[order]

    # convert to datetimes
    t     = Time(bjd, format='jd', scale='tdb')
    dates = t.to_datetime()

    # ---------- STATS ----------
    primary_amp  = (np.nanmax(rv1) - np.nanmin(rv1)) / 2.0
    primary_mean = np.nanmean(rv1)
    primary_std  = np.nanstd(rv1)
    rv1_hi_max   = np.nanmax(rv1_hi)
    rv1_hi_mean  = np.nanmean(rv1_hi)
    rv1_lo_max   = np.nanmax(rv1_lo)
    rv1_lo_mean  = np.nanmean(rv1_lo)

    print(f"""{star} — Primary (combined)
  Semi-Amplitude  : {primary_amp:.3f} km/s
  Mean            : {primary_mean:.3f} km/s
  STD             : {primary_std:.3f} km/s
Primary Error Bars
  RV1_hi max      : {rv1_hi_max:.3f} km/s
  RV1_hi mean     : {rv1_hi_mean:.3f} km/s
  RV1_lo max      : {rv1_lo_max:.3f} km/s
  RV1_lo mean     : {rv1_lo_mean:.3f} km/s
""")

    mask2 = ~np.isnan(rv2)

    if mask2.any():
        rv2_v    = rv2[mask2]
        rv2_lo_v = rv2_lo[mask2]
        rv2_hi_v = rv2_hi[mask2]
        dates_v  = dates[mask2]

        secondary_amp  = (np.nanmax(rv2_v) - np.nanmin(rv2_v)) / 2.0
        secondary_mean = np.nanmean(rv2_v)
        secondary_std  = np.nanstd(rv2_v)
        rv2_hi_max     = np.nanmax(rv2_hi_v)
        rv2_hi_mean    = np.nanmean(rv2_hi_v)
        rv2_lo_max     = np.nanmax(rv2_lo_v)
        rv2_lo_mean    = np.nanmean(rv2_lo_v)

        print(f"""{star} — Secondary (combined)
  Semi-Amplitude  : {secondary_amp:.3f} km/s
  Mean            : {secondary_mean:.3f} km/s
  STD             : {secondary_std:.3f} km/s
Secondary Error Bars
  RV2_hi max      : {rv2_hi_max:.3f} km/s
  RV2_hi mean     : {rv2_hi_mean:.3f} km/s
  RV2_lo max      : {rv2_lo_max:.3f} km/s
  RV2_lo mean     : {rv2_lo_mean:.3f} km/s
""")
    else:
        secondary_amp  = np.nan
        secondary_mean = np.nan
        print(f"{star} — Secondary is all NaN in combined data.\n")

    # ---------- PLOT ----------
    fig, ax = plt.subplots(figsize=(7, 4))

    # primary
    ax.errorbar(dates, rv1,
                yerr=[rv1_lo, rv1_hi],
                fmt='o',
                label=f"Primary\n(K={primary_amp:.2f} km/s)",
                ecolor='grey', capsize=1, elinewidth=0.5,
                ms=5, alpha=0.9)

    # secondary
    if mask2.any():
        ax.errorbar(dates_v, rv2_v,
                    yerr=[rv2_lo_v, rv2_hi_v],
                    fmt='x',
                    label=f"Secondary\n(K={secondary_amp:.2f} km/s)",
                    ecolor='grey', capsize=1, elinewidth=0.5,
                    ms=5, alpha=0.9)
        ax.axhline(secondary_mean, alpha=0.25)

    # primary mean
    ax.axhline(primary_mean, alpha=0.25)

    # y padding
    if mask2.any():
        y_min = min(np.nanmin(rv1), np.nanmin(rv2_v)) - 5
        y_max = max(np.nanmax(rv1), np.nanmax(rv2_v)) + 5
    else:
        y_min = np.nanmin(rv1) - 5
        y_max = np.nanmax(rv1) + 5
    ax.set_ylim(y_min, y_max)

    # date formatting
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
    fig.autofmt_xdate()

    ax.set_xlabel("Date (UTC)")
    ax.set_ylabel("Radial Velocity (km/s)")
    ax.set_title(f"RV vs Time (combined) for {star}")

    # legend outside
    box = ax.get_position()
    ax.set_position([box.x0, box.y0, box.width * 0.75, box.height])
    ax.legend(loc="center left", bbox_to_anchor=(1, 0.5))

    plt.tight_layout()
    #plt.savefig(os.path.join(outdir_comb, f"{star}_combined.png"), dpi=135, bbox_inches="tight")
    plt.show()

    print(f"{star}: combined points = {len(bjd)}\n")
```
